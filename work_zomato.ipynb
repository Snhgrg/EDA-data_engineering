```python
df_country_code = pd.read_excel('Country-Code.xlsx')
```

```python
df_country_code.head()
```

```python
final_df = pd.merge(df, df_country_code , on='Country Code' , how='left')
```

```python
final_df.head()
```

```python
final_df.Country.value_counts()
```

```python
Countries = final_df.Country.value_counts().index
print(Countries)
```

```python
Counties_val = final_df.Country.value_counts().values
print(Counties_val)
```

```python
plt.pie(Counties_val , labels= Countries)
```

```python
#for top 3 countries
plt.pie(Counties_val[:3] , labels= Countries[:3] ,autopct='%1.2f%%')
```

```python
final_df.groupby(['Has Table booking']).size()
# or final_df['Has Table booking'].value_counts
```

```python
final_df.groupby(['Aggregate rating' , 'Rating color' , 'Rating text']).size().reset_index()
```

```python
ratings= final_df.groupby(['Aggregate rating' , 'Rating color' , 'Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
ratings
```

#Observation
#When Rating is between 4.5 to 4.9 the rating is excellent
#When Rating is between 4.0 to 4.4 the rating is very good

```python
sns.barplot(x='Rating color' , y='Rating Count' ,data=ratings , estimator=sum) # estimator default=avg.
```

Observation - Most Ratings lie in Orange Zone

```python
import matplotlib
matplotlib.rcParams['figure.figsize']=(11,4)
sns.barplot(x='Aggregate rating' , y='Rating Count' ,hue='Rating text' , data=ratings)
```

Observation
1) Not Rated count is very high
2) Most of the Rating are between 2.5 to 3.4
